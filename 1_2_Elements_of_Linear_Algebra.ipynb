## 1.2.1 Linear Spaces

In order for a vector to be part of a vector space it must be closed under addition and multiplication by a scalar.

**1.2.1.1 Linear Combinations**

*   A linear combination is a vector that is created by multiplying each vector within a subset by a constant and adding the results.

*   **Defintion 1.2.1 Linear Subspace**

    *   "a subset that is closed under vector addition and scalar multpliocation."

The following vectors are part of the vector space of R^3 (space of all real 3D vectors)

```python
# vector definitions
u = np.array([2,4,-6])
v = np.array([8,-4,2])
print(u)
print(v)
```

Example of a linear combination



```python
# linear combination of the two vectors by a constant and adding
lc = (-2) * u + 5 * v
print("-2u + 5v = ",lc)
```

Showing closure under vector addition and scalar multiplication

```python
print("Vector addition: \nu + v=",u+v)
print("\nScalar multiplication:")
print("3.14 * v=",3.14*v)
print("3.14 * u=",3.14*u)
```

A vector within the given vector space should also prove the existence of a zero vector in the subspace as well as the additive inverse (v + (-v) = 0)

```python
# zero vector
print("Zero Vector = ",np.zeros_like(u))

# additive inverse
print("Additive Inverse (u) = ", u + (-u))
```

Since the resulting vectors still reside within the R^3 subspace of all real 3D vectors, then our vectors u and v are proven to be within the vector space R^3.

## 1.2.2 Orthogonality

Orthogonality is when two vectors lie on perpendicular planes. The angle between these vectors will always be 90 degrees.

If two vectors have a dot product of zero, they are orthogonal.

```python
# define vectors
u = np.array([2,-1,2])
v = np.array([-2,2,3])
b = np.array([-2,2,-2])

udv = np.dot(u,v)

# compute dot product
print("u dot v = ", udv)
print("u dot b = ", np.dot(u,b))
print("v dot b = ", np.dot(v,b))
```

Here it can be seen that u dot v has a product of zero, but u dot b and v dot b do not. Therefore, vectors u and v are orthogonal to each other while b is not orthogonal to either.

This can be confirmed by computing the angle between the two vectors as the angle between any two orthogonal vectors will be 90 degrees.


```python
a = np.degrees(np.arccos(udv / np.linalg.norm(u) * np.linalg.norm(v)))
print("Angle between vectors u and v = ", a)
```

Since these vectors are defined in three dimensions, they must be displayed on a 3-dimenional graph.

```python
fig = plt.figure(figsize=(5, 5))
g = fig.add_subplot(111, projection='3d')
g.set(xlim=(-5,5),ylim=(-5,5),zlim=(-5,5))
g.quiver(0, 0, 0, u[0], u[1], u[2], color='green')
g.quiver(0, 0, 0, v[0], v[1], v[2], color='purple')
plt.show()
```
